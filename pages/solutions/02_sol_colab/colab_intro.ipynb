## Sumowanie ciągu geometrycznego

Ciąg geometryczny to ciąg liczb, w którym każdy kolejny wyraz różni się od poprzedniego stałą wielokrotnością, zwaną ilorazem ciągu $q$.

Wzór na $n$-ty wyraz ciągu geometrycznego o pierwszym wyrazie $a_1$ i ilorazie $q$ to:
$$a_n = a_1 \cdot q^{n-1}$$

### Wzór na sumę $n$ pierwszych wyrazów
Suma $S_n$ pierwszych $n$ wyrazów ciągu geometrycznego wyraża się wzorem:

* Dla $q \neq 1$:
$$S_n = a_1 \frac{1 - q^n}{1 - q}$$

* Dla $q = 1$:
$$S_n = a_1 \cdot n$$

```python
import matplotlib.pyplot as plt

k = list(range(1, 21))
y = [i**2 for i in k]

plt.figure(figsize=(8, 5))
plt.plot(k, y, marker='o', linestyle='-', color='b', label='$y = k^2$')
plt.title(r'Wykres funkcji $y = k^2$ dla $k = 1, \dots, 20$')
plt.xlabel('k')
plt.ylabel('y')
plt.xticks(k)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.savefig('wykres.png', dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

g = 9.81
L = 1.0
theta0 = np.radians(45)
dt = 0.04
t_max = 5.0
t = np.arange(0, t_max, dt)

omega = np.sqrt(g / L)
theta = theta0 * np.cos(omega * t)
x = L * np.sin(theta)
y = -L * np.cos(theta)

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_title('Animacja Wahadła')

line, = ax.plot([], [], 'o-', lw=3, color='blue', markersize=10)
trace, = ax.plot([], [], '-', lw=1, color='red', alpha=0.6)

def init():
    line.set_data([], [])
    trace.set_data([], [])
    return line, trace

def update(frame):
    line.set_data([0, x[frame]], [0, y[frame]])
    trace.set_data(x[:frame], y[:frame])
    return line, trace

ani = FuncAnimation(fig, update, frames=len(t), init_func=init, blit=True, interval=int(dt*1000))
ani.save('video.mp4', writer='ffmpeg', fps=int(1/dt), codec='libx264', extra_args=['-pix_fmt', 'yuv420p'])
plt.close(fig)
print('Zapisano video.mp4')
```

```python
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

t = np.linspace(0, 2 * np.pi, 500)
x = np.sin(t) * np.cos(t)
y = np.sin(t) * np.sin(t)
z = np.cos(t)

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, label='Parametric Curve')
ax.set_title('3D Parametric Curve', fontsize=14)
ax.set_xlabel('x(t)')
ax.set_ylabel('y(t)')
ax.set_zlabel('z(t)')
ax.legend()
plt.savefig('parametric_curve.png', dpi=200)
plt.show()
```
