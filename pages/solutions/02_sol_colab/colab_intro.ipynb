## Sumowanie ciągu geometrycznego

Ciąg geometryczny to ciąg liczb, w którym każdy kolejny wyraz różni się od poprzedniego stałą wielokrotnością, zwaną ilorazem ciągu $q$.

Wzór na $n$-ty wyraz ciągu geometrycznego o pierwszym wyrazie $a_1$ i ilorazie $q$ to:
$$a_n = a_1 \cdot q^{n-1}$$

### Wzór na sumę $n$ pierwszych wyrazów
Suma $S_n$ pierwszych $n$ wyrazów ciągu geometrycznego wyraża się wzorem:

* Dla $q \neq 1$:
$$S_n = a_1 \frac{1 - q^n}{1 - q}$$

* Dla $q = 1$:
$$S_n = a_1 \cdot n$$

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Parametry wahadła i symulacji
g = 9.81  # przyspieszenie ziemskie [m/s^2]
L = 1.0   # długość wahadła [m]
theta0 = np.radians(45)  # maksymalne wychylenie początkowe (45 stopni)
dt = 0.04  # krok czasowy
t_max = 5.0  # czas trwania symulacji [s]
t = np.arange(0, t_max, dt)

# Rozwiązanie przybliżone
omega = np.sqrt(g / L)
theta = theta0 * np.cos(omega * t)

# Współrzędne końca wahadła
x = L * np.sin(theta)
y = -L * np.cos(theta)

# Inicjalizacja wykresu
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_title("Animacja Wahadła")

# Elementy rysunku
line, = ax.plot([], [], 'o-', lw=3, color='blue', markersize=10) # Ramie i masa
trace, = ax.plot([], [], '-', lw=1, color='red', alpha=0.6) # Ślad ruchu

# Funkcja inicjalizująca animację
def init():
    line.set_data([], [])
    trace.set_data([], [])
    return line, trace

# Funkcja aktualizująca każdą klatkę
def update(frame):
    line.set_data([0, x[frame]], [0, y[frame]])
    trace.set_data(x[:frame], y[:frame])
    return line, trace

# Konfiguracja animacji
ani = FuncAnimation(fig, update, frames=len(t), init_func=init, blit=True, interval=dt*1000)

# Zapis do pliku mp4 na dysku
ani.save('video.mp4', writer='ffmpeg', fps=int(1/dt), codec='libx264', extra_args=['-pix_fmt', 'yuv420p'])

plt.close()

# Wyświetlenie interaktywnego odtwarzacza JSHTML w Colabie
plt.rc('animation', html='jshtml')
ani
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parametric equations
t = np.linspace(0, 10 * np.pi, 1000)

x = np.cos(t)
y = np.sin(t)
z = t

# 3D Plot
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, linewidth=2, label="Helix")

ax.set_title("3D Helix", fontsize=14)
ax.set_xlabel(r"$x(t)=\cos(t)$")
ax.set_ylabel(r"$y(t)=\sin(t)$")
ax.set_zlabel(r"$z(t)=t$")

ax.legend()
plt.savefig('parametric_curve.png', dpi=200)
plt.show()
```
